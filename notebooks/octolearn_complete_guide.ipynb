# 5-Minute Quickstart: The Simplest Way to Use OctoLearn

## Step 1: Load Your Data

````python
# Load the famous Titanic dataset
from seaborn import load_dataset

titanic = load_dataset('titanic')
print("📊 Dataset loaded!")
print(f"Shape: {titanic.shape}")
print(f"\nFirst few rows:")
print(titanic.head())
print(f"\nData types:\n{titanic.dtypes}")
print(f"\nMissing values:\n{titanic.isnull().sum()}")
````

## Step 2: Prepare Your Data

Prepare features (X) and target (y):

````python
# Separate features and target
y = titanic['survived']  # Target: Did the passenger survive?
X = titanic.drop('survived', axis=1)

print(f"✅ Data prepared!")
print(f"Features (X): {X.shape}")
print(f"Target (y): {y.shape}")
print(f"\nTarget distribution:")
print(y.value_counts())
````

## Step 3: Run OctoLearn (One Line!)

That's it! Just two lines of code:

````python
print("🚀 Starting OctoLearn Pipeline...")
print("=" * 60)

# Line 1: Create AutoML instance
automl = AutoML(show_progress=True)

# Line 2: Fit the data
automl.fit(X, y)
````

## Step 4: Get Insights

Access the results with simple methods:

````python
# Get dataset profile
profile = automl.report()
print("📈 Dataset Profile:")
print(f"  • Task Type: {profile['task_type']}")
print(f"  • Rows: {profile['n_rows']}")
print(f"  • Columns: {profile['n_columns']}")
print(f"  • Numeric Features: {profile['numeric_features']}")
print(f"  • Categorical Features: {profile['categorical_features']}")
print(f"  • Missing Values: {profile['missing_values']}")
print(f"  • Duplicates: {profile['duplicates']}")

# Get risk score
risk = automl.get_risk_score()
print(f"\n⚠️ Data Quality Risk Score: {risk['score']:.1f}/100")
print(f"   Category: {risk['category']}")

# Get preprocessing suggestions
suggestions = automl.get_preprocessing_suggestions()
print(f"\n🔧 Preprocessing Suggestions:")
for i, suggestion in enumerate(suggestions[:3], 1):
    print(f"   {i}. {suggestion}")
````

# Understanding the 4 Phases

OctoLearn runs 4 interconnected phases automatically:

## Phase 1: 📊 Dataset Profiling
**What it does:** Analyzes your entire dataset and extracts 16 intelligent metrics

**Key Metrics:**
- Row & column counts
- Data types distribution
- Missing values analysis
- Duplicate detection
- Class imbalance detection
- Feature correlations
- Task type classification (classification vs regression)

## Phase 2: 🔍 Exploratory Data Analysis  
**What it does:** Identifies patterns, anomalies, and actionable insights

**Features:**
- Outlier detection (IQR, Isolation Forest, Z-score methods)
- Feature interaction analysis (polynomial, pairwise, ratio interactions)
- Correlation analysis
- Risk scoring (0-100 scale)
- Strategic recommendations
- Preprocessing suggestions

## Phase 3: 🧹 Automatic Data Cleaning & Preprocessing
**What it does:** Cleans and prepares data automatically

**Actions:**
- Removes duplicates
- Identifies and removes ID columns
- Removes constant/low-variance columns
- Handles missing values (mean/median/KNN for numeric, mode for categorical)
- Encodes categorical variables
- Scales numeric features

## Phase 4: 🤖 Model Training & Optimization
**What it does:** Trains multiple models with automatic hyperparameter tuning

**Models Trained (Classification):**
- Logistic Regression
- Random Forest
- Gradient Boosting
- XGBoost
- LightGBM
- Support Vector Machines

**Optimization:**
- Optuna hyperparameter optimization
- 50 trials per model
- 5-fold cross-validation
- Model registry for versioning

# Detailed Deep Dive: What's Happening Behind the Scenes

Let's explore each phase in detail:

## Phase 1 Deep Dive: Dataset Profiling

Let's examine the profile in detail:

````python
print("📊 PHASE 1: Complete Dataset Profile")
print("=" * 60)

profile = automl.report()

# Display all available profile metrics
profile_items = [
    ('Task Type', profile.get('task_type')),
    ('Total Rows', profile.get('n_rows')),
    ('Total Columns', profile.get('n_columns')),
    ('Numeric Features', profile.get('numeric_features')),
    ('Categorical Features', profile.get('categorical_features')),
    ('Datetime Features', profile.get('datetime_features')),
    ('Total Missing Values', profile.get('missing_values')),
    ('Duplicate Rows', profile.get('duplicates')),
    ('Memory Usage (MB)', round(profile.get('memory_usage', 0) / 1024 / 1024, 2)),
    ('Class Imbalance Detected', profile.get('class_imbalance_detected')),
    ('Data Leakage Suspects', profile.get('leakage_suspects')),
    ('Skewness Range', f"{profile.get('skewness_min', 0):.2f} to {profile.get('skewness_max', 0):.2f}"),
]

for label, value in profile_items:
    print(f"  {label:.<35} {value}")
````

## Phase 2 Deep Dive: Exploratory Data Analysis

### Outlier Detection Results:

````python
print("🔍 PHASE 2: Exploratory Data Analysis")
print("=" * 60)

# Outlier Analysis
outliers = automl.get_outlier_analysis()
if outliers:
    print("\n🎯 Outlier Detection:")
    for method, indices in outliers.items():
        print(f"  • {method}: {len(indices) if isinstance(indices, (list, np.ndarray)) else 0} outliers detected")

# Risk Assessment
risk = automl.get_risk_score()
print(f"\n⚠️ Data Quality Risk Assessment:")
print(f"  • Overall Risk Score: {risk['score']:.1f}/100")
print(f"  • Risk Category: {risk['category']}")
print(f"  • Risk Factors:")
for factor in risk.get('factors', [])[:5]:
    print(f"    - {factor}")

# Feature Importance
try:
    importance = automl.get_feature_importance()
    print(f"\n⭐ Feature Importance Ranking:")
    if isinstance(importance, dict):
        for i, (feature, score) in enumerate(list(importance.items())[:5], 1):
            print(f"  {i}. {feature}: {score:.4f}")
except:
    print("\n⭐ Feature Importance: Not available in this phase")
````

## Phase 3 Deep Dive: Automatic Data Cleaning

What was cleaned and prepared:

````python
print("🧹 PHASE 3: Automatic Data Cleaning & Preprocessing")
print("=" * 60)

# Cleaning log
cleaning_log = automl.get_cleaning_log()
print(f"\nOriginal dataset: {X.shape}")
print(f"Cleaned dataset: {(len(automl.X_), automl.X_.shape[1]) if automl.X_ is not None else 'Not available'}")

if cleaning_log:
    print(f"\n🔧 Cleaning Actions Taken:")
    for action, details in cleaning_log.items():
        if isinstance(details, dict):
            print(f"  • {action}: {details.get('count', 'N/A')} items")
        else:
            print(f"  • {action}: {details}")
else:
    print("  (Cleaning details available in automl.cleaning_log_)")

# Preprocessing suggestions
suggestions = automl.get_preprocessing_suggestions()
print(f"\n💡 Recommended Preprocessing Steps:")
if isinstance(suggestions, (list, dict)):
    for i, suggestion in enumerate(list(suggestions)[:5] if isinstance(suggestions, (list, dict)) else [], 1):
        print(f"  {i}. {suggestion}")
else:
    print("  (Check automl.get_preprocessing_suggestions() for details)")
````

## Phase 4 Deep Dive: Model Training & Optimization

Now let's train models with automatic hyperparameter optimization:

````python
print("🤖 PHASE 4: Model Training & Hyperparameter Optimization")
print("=" * 60)

# Train models
print("\n⏳ Training multiple models with Optuna optimization...")
training_results = automl.train_auto_models()

if 'error' not in training_results:
    print("\n✅ Model Training Complete!")
    print(f"\n🏆 Best Model: {training_results.get('best_model', 'N/A')}")
    print(f"   Score: {training_results.get('best_score', 0):.4f}")
    
    print("\n📊 All Models Performance:")
    results_dict = training_results.get('results', {})
    if results_dict:
        for model_name, score in list(results_dict.items())[:5]:
            print(f"  • {model_name}: {score:.4f}")
else:
    print(f"⚠️ Training error: {training_results['error']}")

# Get best model
best_model = automl.get_best_model()
if best_model:
    print(f"\n✨ Best Model Object: {type(best_model).__name__}")
    print(f"   Ready for predictions!")
else:
    print("⚠️ No trained model available")
````

# Advanced Usage: Customizing OctoLearn

## Example 1: Disable Specific Phases

You have fine-grained control over which phases to run:

````python
# Example: Skip outlier detection and model training, focus on profiling

print("Example 1: Custom Configuration (Skip Outliers & Training)")
print("=" * 60)

# Create AutoML with custom parameters
automl_custom = AutoML(
    detect_outliers=False,      # Skip outlier detection
    train_models=False,         # Don't train models, just analyze
    analyze_interactions=True,  # Still analyze interactions
    auto_clean=True,            # Still clean data
    show_progress=True
)

# Fit with custom config
automl_custom.fit(X, y)

print("\n✅ Completed analysis without outlier detection and model training!")
print(f"   Profile available: {automl_custom.report() is not None}")
print(f"   Outlier analysis: {automl_custom.get_outlier_analysis() or 'Skipped'}")
print(f"   Best model: {automl_custom.get_best_model() or 'Not trained'}")
````

## Example 2: Use Full Data vs Sampling

For large datasets, control memory usage:

````python
print("Example 2: Data Sampling for Speed vs Full Data")
print("=" * 60)

# Option A: Use full data (slower but more accurate)
config_full = {
    'use_full_data': True,
    'show_progress': False  # Quiet mode for comparison
}

# Option B: Sample for speed (default, good for large datasets)
config_sampled = {
    'use_full_data': False,
    'sample_size': 300,  # Use only 300 rows
    'show_progress': False
}

print(f"\nConfiguration Options:")
print(f"  Full Data:     use_full_data=True")
print(f"  Sampled:       use_full_data=False, sample_size=300")
print(f"\nFor Titanic ({X.shape[0]} rows):")
print(f"  • Full data processing: Uses all 891 rows")
print(f"  • Sampling (default):   Uses 300-500 rows (faster!)")
````

# Complete Real-World Example: Predict Titanic Survival

Let's build a complete ML pipeline from scratch to predict who survived the Titanic.

## The Full Workflow (10 lines of code!)

````python
print("🎯 COMPLETE REAL-WORLD EXAMPLE: Titanic Survival Prediction")
print("=" * 70)

# Step 1: Load data
from seaborn import load_dataset
df = load_dataset('titanic')
print(f"\n1️⃣ Data loaded: {df.shape[0]} passengers, {df.shape[1]} features")

# Step 2: Prepare features and target
y = df['survived']
X = df.drop('survived', axis=1)
print(f"2️⃣ Data prepared: {X.shape[0]} samples, {X.shape[1]} features")

# Step 3: Initialize AutoML with production settings
automl_prod = AutoML(
    show_progress=False,  # Production mode
    generate_recommendations=True,
    n_models=5  # Train top 5 models
)
print(f"3️⃣ AutoML initialized with production settings")

# Step 4: Fit the complete pipeline
automl_prod.fit(X, y)
print(f"4️⃣ Pipeline fitted (all 4 phases complete)")

# Step 5: Get insights
profile = automl_prod.report()
print(f"\n5️⃣ Dataset Profile:")
print(f"   ├─ Task: {profile['task_type']}")
print(f"   ├─ Numeric features: {profile['numeric_features']}")
print(f"   ├─ Categorical features: {profile['categorical_features']}")
print(f"   └─ Missing values: {profile['missing_values']}")

# Step 6: Check data quality
risk = automl_prod.get_risk_score()
print(f"\n6️⃣ Data Quality: {risk['category']} (Score: {risk['score']:.0f}/100)")

# Step 7: Get preprocessing recommendations
suggestions = automl_prod.get_preprocessing_suggestions()
print(f"\n7️⃣ System Recommendations:")
for i, suggestion in enumerate(list(suggestions)[:3], 1) if isinstance(suggestions, (list, dict)) else []:
    print(f"   {i}. {suggestion}")

# Step 8: Train models with optimization
results = automl_prod.train_auto_models()
if 'error' not in results:
    print(f"\n8️⃣ Models Trained:")
    print(f"   🏆 Best Model: {results.get('best_model')}")
    print(f"   ⭐ Score: {results.get('best_score', 0):.4f}")
else:
    print(f"\n8️⃣ Training results: Check logs")

# Step 9: Get the best model
best_model = automl_prod.get_best_model()
print(f"\n9️⃣ Best Model Ready: {type(best_model).__name__}")

# Step 10: Make predictions
print(f"\n🔟 Making Predictions on First 5 Passengers:")
predictions = best_model.predict(X.head())
print(f"   Predictions: {predictions}")
print(f"   {sum(predictions)} out of 5 predicted to survive")

print(f"\n" + "=" * 70)
print("✅ COMPLETE PIPELINE FINISHED IN ~10 LINES OF CODE!")
print("=" * 70)
````
