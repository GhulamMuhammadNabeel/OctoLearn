## Installation & Setup

````python
import sys
import warnings
warnings.filterwarnings('ignore')

# Verify OctoLearn installation
try:
    import octolearn
    print(f"✅ OctoLearn {octolearn.__version__} installed")
except ImportError:
    print("❌ OctoLearn not found. Installing...")
    import subprocess
    subprocess.check_call([sys.executable, "-m", "pip", "install", "octolearn"])

# Import everything needed
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from octolearn import AutoML

print(f"\n✅ All imports successful!")
print(f"   Python: {sys.version.split()[0]}")
print(f"   Pandas: {pd.__version__}")
print(f"   NumPy: {np.__version__}")
````

---
## Part 1: Load & Explore Sample Data

We'll demonstrate using multiple datasets from seaborn and sklearn to show OctoLearn works with any data.

````python
# Option A: Use Titanic dataset (classification)
from seaborn import load_dataset

titanic = load_dataset('titanic')
print("📊 Dataset loaded: Titanic")
print(f"Shape: {titanic.shape}")
print(f"\nColumns: {titanic.columns.tolist()}")
print(f"\nFirst rows:")
print(titanic.head())
print(f"\nMissing values:")
print(titanic.isnull().sum())
print(f"\nData types:")
print(titanic.dtypes)
````

````python
# Prepare data for OctoLearn
y = titanic['survived']  # Target: Did passenger survive?
X = titanic.drop('survived', axis=1)

print(f"✅ Data prepared!")
print(f"Features (X): {X.shape}")
print(f"Target (y): {y.shape}")
print(f"\nTarget distribution:")
print(y.value_counts())
print(f"\nTask: Binary Classification")
````

---
## Part 2: Quick Start (Automatic Everything)

Just 2 lines to run complete ML pipeline!

````python
# Complete AutoML pipeline with 1 line (defaults handle everything)
automl = AutoML(show_progress=True)  # All smart defaults

# Fit the complete pipeline (4 phases)
automl.fit(X, y)

print("\n" + "="*60)
print("✅ COMPLETE PIPELINE FINISHED!")
print("="*60)
````

---
## Part 3: PHASE 1 - Dataset Profiling (16 Metrics)

````python
print("📊 PHASE 1: Complete Dataset Profile")
print("="*60)

profile = automl.report()

# Display all profile metrics
profile_display = [
    ('Task Type', profile.get('task_type')),
    ('Total Rows', profile.get('n_rows')),
    ('Total Columns', profile.get('n_columns')),
    ('Numeric Features', profile.get('numeric_features')),
    ('Categorical Features', profile.get('categorical_features')),
    ('Datetime Features', profile.get('datetime_features')),
    ('Total Missing Values', profile.get('missing_values')),
    ('Duplicate Rows', profile.get('duplicates')),
    ('Memory Usage (MB)', round(profile.get('memory_usage', 0) / 1024 / 1024, 2)),
    ('Class Imbalance Detected', profile.get('class_imbalance_detected')),
]

for label, value in profile_display:
    print(f"  {label:.<35} {value}")

print(f"\n💡 Profile Dictionary Keys: {list(profile.keys())[:10]}...")
````

---
## Part 4: PHASE 2 - Data Quality Analysis & Risk Assessment

````python
print("🔍 PHASE 2: Data Quality Analysis")
print("="*60)

# Risk Score (0-100)
risk = automl.get_risk_score()
print(f"\n⚠️ DATA QUALITY RISK SCORE")
print(f"  Score: {risk['score']:.1f}/100")
print(f"  Category: {risk['category']}")
if 'factors' in risk and risk['factors']:
    print(f"  Risk Factors:")
    for factor in risk.get('factors', [])[:5]:
        print(f"    • {factor}")

# Outlier Detection
print(f"\n🎯 OUTLIER DETECTION")
outliers = automl.get_outlier_analysis()
if outliers:
    for method, count in outliers.items():
        if isinstance(count, (list, np.ndarray)):
            print(f"  • {method}: {len(count)} outliers detected")
        elif isinstance(count, int):
            print(f"  • {method}: {count} outliers detected")

# Feature Importance
print(f"\n⭐ FEATURE IMPORTANCE RANKING")
try:
    importance = automl.get_feature_importance()
    if isinstance(importance, dict) and importance:
        for i, (feature, score) in enumerate(list(importance.items())[:5], 1):
            print(f"  {i}. {feature:.<20} {score:.4f}")
except:
    print("  (Available after model training)")
````

---
## Part 5: PHASE 3 - Automatic Data Cleaning & Preprocessing

````python
print("🧹 PHASE 3: Automatic Data Cleaning")
print("="*60)

# Original vs Cleaned data
print(f"\n📦 DATA SHAPE CHANGES")
print(f"  Original: {X.shape[0]} rows × {X.shape[1]} columns")
if hasattr(automl, 'X_') and automl.X_ is not None:
    print(f"  Cleaned:  {automl.X_.shape[0]} rows × {automl.X_.shape[1]} columns")
    print(f"  Removed:  {X.shape[0] - automl.X_.shape[0]} rows, {X.shape[1] - automl.X_.shape[1]} columns")

# Cleaning log
cleaning_log = automl.get_cleaning_log()
if cleaning_log:
    print(f"\n🔧 CLEANING ACTIONS TAKEN")
    for action, details in list(cleaning_log.items())[:8]:
        if isinstance(details, dict):
            print(f"  • {action}: {details}")
        else:
            print(f"  • {action}: {details}")

# Preprocessing suggestions
print(f"\n💡 PREPROCESSING RECOMMENDATIONS")
suggestions = automl.get_preprocessing_suggestions()
if isinstance(suggestions, (list, dict)):
    for i, suggestion in enumerate(list(suggestions)[:5] if isinstance(suggestions, (list, dict)) else [], 1):
        print(f"  {i}. {suggestion}")
````

---
## Part 6: Access Cleaned & Preprocessed Data

**KEY FEATURE:** Users can access data at any stage of the pipeline!

````python
print("📂 DATA ACCESS AT ALL STAGES")
print("="*60)

# 1. Original data
print(f"\n1️⃣ ORIGINAL DATA (untouched)")
if hasattr(automl, 'X_original_'):
    print(f"  automl.X_original_.shape = {automl.X_original_.shape}")
    print(f"  automl.y_original_.shape = {automl.y_original_.shape}")
    print(f"  Example:\n{automl.X_original_.head(2)}")

# 2. Sampled data (if sampling enabled)
print(f"\n2️⃣ SAMPLED DATA (for profiling)")
if hasattr(automl, 'X_'):
    print(f"  automl.X_.shape = {automl.X_.shape}")
    print(f"  automl.y_.shape = {automl.y_.shape}")
    print(f"  Example:\n{automl.X_.head(2)}")

# 3. Cleaned data (after auto-cleaning)
print(f"\n3️⃣ CLEANED DATA (after Phase 3)")
if hasattr(automl, 'X_') and automl.X_ is not None:
    print(f"  Shape: {automl.X_.shape}")
    print(f"  Columns: {automl.X_.columns.tolist()[:5]}...")
    print(f"  Data types: ")
    print(f"    Numeric: {automl.X_.select_dtypes(include=['int64', 'float64']).shape[1]}")
    print(f"    Categorical: {automl.X_.select_dtypes(include=['object']).shape[1]}")
````

---
## Part 7: PHASE 4 - Model Training with Optuna HPO

````python
print("🤖 PHASE 4: Model Training & Hyperparameter Optimization")
print("="*60)

# Train multiple models
print("\n⏳ Training 5 models with Optuna optimization...")
print("   Each model gets 50 trials with 5-fold cross-validation")

training_results = automl.train_auto_models()

if 'error' not in training_results:
    print("\n✅ MODEL TRAINING COMPLETE!")
    
    # Best model
    print(f"\n🏆 BEST MODEL")
    print(f"  Name: {training_results.get('best_model', 'N/A')}")
    print(f"  Score: {training_results.get('best_score', 0):.4f}")
    
    # All trained models
    print(f"\n📊 ALL MODELS PERFORMANCE")
    results_dict = training_results.get('results', {})
    if results_dict:
        for i, (model_name, score) in enumerate(list(results_dict.items())[:6], 1):
            print(f"  {i}. {model_name:.<25} Score: {score:.4f}")
else:
    print(f"⚠️ Training info: {training_results}")
````

---
## Part 8: Get Trained Models & Make Predictions

````python
print("🎯 MODEL ACCESS & PREDICTIONS")
print("="*60)

# Get best model
best_model = automl.get_best_model()
if best_model:
    print(f"\n✨ Best Model Object: {type(best_model).__name__}")
    print(f"   Ready for predictions!")
    
    # Make predictions
    print(f"\n🔮 PREDICTIONS ON FIRST 10 ROWS")
    try:
        preds = best_model.predict(X.head(10))
        print(f"  Predictions: {preds}")
        print(f"  Type: {type(preds[0])}")
    except Exception as e:
        print(f"  Note: Predictions require preprocessed data")
else:
    print(f"⚠️ No model trained yet (set train_models=True)")
````

---
## Part 9: ADVANCED - Customized Pipeline Control

````python
print("⚙️ ADVANCED: Customize Pipeline Parameters")
print("="*60)

# Example 1: Skip outlier detection, focus on profiling
print("\n1️⃣ CUSTOM CONFIG: EDA Only (No Model Training)")
autom_eda = AutoML(
    detect_outliers=False,      # Skip outlier detection
    train_models=False,         # Don't train models
    analyze_interactions=True,  # Still analyze interactions
    auto_clean=True,            # Still clean data
    show_progress=False
)
autom_eda.fit(X, y)
print("✅ Completed EDA without model training and outlier detection")
print(f"   Profile: {autom_eda.report() is not None}")
print(f"   Best model: {autom_eda.get_best_model() or 'Not trained'}")

# Example 2: Full data processing
print(f"\n2️⃣ CUSTOM CONFIG: Use Full Data (Better Accuracy)")
autom_full = AutoML(
    use_full_data=True,         # Use all data
    generate_shap=False,        # Skip expensive SHAP
    show_progress=False
)
autom_full.fit(X, y)
print(f"✅ Using full dataset: {autom_full.X_.shape[0]} rows")

# Example 3: Sampling for speed
print(f"\n3️⃣ CUSTOM CONFIG: Fast Prototyping (Sampling)")
autom_fast = AutoML(
    use_full_data=False,        # Sample data
    sample_size=200,            # Use only 200 rows
    generate_shap=False,        # Skip SHAP
    show_progress=False
)
autom_fast.fit(X, y)
print(f"✅ Fast mode: {autom_fast.X_.shape[0]} rows sampled")
````

---
## Part 10: Feature Interaction Analysis

````python
print("🔗 FEATURE INTERACTIONS")
print("="*60)

interactions = automl.get_interaction_analysis()
if interactions:
    print(f"\n✨ Interaction Types Detected:")
    for itype, details in interactions.items():
        if isinstance(details, dict) and 'n_interactions' in details:
            print(f"  • {itype}: {details['n_interactions']} interactions")
else:
    print("  (Enable analyze_interactions=True)")
````

## Part 11: Model Registry (Production Feature)

````python
print("📦 MODEL REGISTRY (Version Control for Models)")
print("="*60)

if hasattr(automl, 'registry_') and automl.registry_:
    print(f"\n✅ Registry initialized!")
    print(f"   Storage: {automl.registry_.storage_type}")
    print(f"   Location: {automl.registry_.db_path}")
    print(f"\n   Features:")
    print(f"   • Automatic model versioning")
    print(f"   • Metadata storage (params, performance)")
    print(f"   • Model recovery & comparison")
    print(f"   • Production deployment ready")
else:
    print("   Use use_registry=True to enable")
````

---
## Part 12: Complete Real-World Example (15 lines)

````python
print("🎯 COMPLETE REAL-WORLD ML PIPELINE (15 lines!)")
print("="*70)

# Step 1: Load data
from sklearn.datasets import load_iris
data = load_iris(as_frame=True)
X_iris = data.data
y_iris = data.target

# Step 2: Create AutoML
automl_iris = AutoML(
    show_progress=False,
    n_models=4,
    use_full_data=True
)

# Step 3: Fit pipeline
automl_iris.fit(X_iris, y_iris)

# Step 4: Get insights
profile_iris = automl_iris.report()
risk_iris = automl_iris.get_risk_score()

# Step 5: Train & get results
results_iris = automl_iris.train_auto_models()
best_model_iris = automl_iris.get_best_model()

# Step 6: Make predictions
predictions_iris = best_model_iris.predict(X_iris.head(5))

print(f"\n✅ IRIS DATASET RESULTS:")
print(f"  Dataset: {profile_iris['n_rows']} rows × {profile_iris['n_columns']} columns")
print(f"  Task: {profile_iris['task_type']}")
print(f"  Risk Score: {risk_iris['score']:.0f}/100 ({risk_iris['category']})")
if 'best_model' in results_iris:
    print(f"  Best Model: {results_iris['best_model']}")
    print(f"  Score: {results_iris['best_score']:.4f}")
print(f"  Predictions (first 5): {predictions_iris}")

print(f"\n" + "="*70)
print(f"✅ COMPLETE ML PIPELINE IN LESS THAN 20 LINES OF CODE!")
print(f"="*70)
````
