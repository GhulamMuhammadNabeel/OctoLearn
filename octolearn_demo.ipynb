# Octolearn: Complete Intelligence Pipeline

## 1. Import Required Libraries

```python
import pandas as pd
import numpy as np
from sklearn.datasets import load_iris, load_breast_cancer
from octolearn import AutoML
import json
print("✓ All dependencies imported successfully")
```

```python
import octolearn
(octolearn.__all__)
```

## 2. Load and Explore Dataset

```python
# # Load Iris dataset
# data = load_iris(as_frame=True)
# X = data.data
# y = data.target
df=pd.read_csv('Messy_Employee_dataset.csv')
df.head()
```

```python
X=df.drop(['Email','Phone','Salary'],axis=1)
y=df['Salary']
```

## 3. Initialize Octolearn and Fit

```python
from octolearn.core import DataProfiler
profile = DataProfiler
print(dir(profile))
```

```python
automl = AutoML(use_full_data=True,scaler='standard',evaluation_metric='rmse')
automl.fit(X, y)
```

```python
automl.generate_report()
```

```python
automl.get_feature_importance()
```

## 4. Dataset Risk Score (Data Quality Credit Score)

```python
risk = automl.get_risk_score()

print(f"\n{'='*60}")
print(f"DATASET RISK SCORE: {risk['score']}/100")
print(f"CATEGORY: {risk['category']}")
print(f"{'='*60}")

if risk['factors']:
    print(f"\nRisk Factors Detected:")
    for factor_name, factor_desc in risk['factors'].items():
        print(f"  ⚠ {factor_desc}")
else:
    print("\n✓ No major risk factors detected")
```

## 5. Feature Importance (Baseline Model)

```python
importance = automl.get_feature_importance()

print("\nFeature Importance Scores:")
print("-" * 45)
for i, (feature, score) in enumerate(importance.items(), 1):
    bar = '█' * int(score * 50)
    print(f"{i:2}. {feature:25} {score:.4f} {bar}")
```

## 6. Preprocessing Recommendations

```python
suggestions = automl.get_preprocessing_suggestions()

print("\nPREPROCESSING STRATEGY:")
print("="*60)

for section, sug_list in suggestions.items():
    section_title = section.replace('_', ' ').upper()
    print(f"\n{section_title}:")
    print("-" * 60)
    
    if isinstance(sug_list, list):
        for suggestion in sug_list:
            print(f"  → {suggestion}")
    else:
        print(f"  → {sug_list}")
```

## 7. Generate Complete Intelligence Report (PDF)

```python
print("\nGenerating comprehensive intelligence report...")
pdf_file = automl.generate_report()

print(f"\n✓ Report saved: {pdf_file}")
print(f"\nReport Contents:")
print("  📊 Risk Score Assessment")
print("  📈 Data Quality Analysis")
print("  📉 Feature Distributions")
print("  🔗 Correlation Heatmap")
print("  🎯 SHAP Feature Importance")
print("  🔧 Preprocessing Recommendations")
print("  💡 Strategic Automation Insights")
```

## 8. Advanced: Multiple Datasets Comparison

```python
# Test with different dataset
cancer_data = load_breast_cancer(as_frame=True)
X_cancer = cancer_data.data
y_cancer = cancer_data.target

automl_cancer = AutoML()
automl_cancer.fit(X_cancer, y_cancer)

risk_cancer = automl_cancer.get_risk_score()

print("\nCOMPARISON: Iris vs Breast Cancer Dataset")
print(f"{'='*60}")
print(f"{'Metric':<25} {'Iris':<20} {'Breast Cancer':<20}")
print(f"{'-'*60}")
print(f"{'Rows':<25} {automl.report().n_rows:<20} {automl_cancer.report().n_rows:<20}")
print(f"{'Columns':<25} {automl.report().n_columns:<20} {automl_cancer.report().n_columns:<20}")
print(f"{'Risk Score':<25} {automl.get_risk_score()['score']:<20} {risk_cancer['score']:<20}")
print(f"{'Risk Category':<25} {automl.get_risk_score()['category']:<20} {risk_cancer['category']:<20}")
```
